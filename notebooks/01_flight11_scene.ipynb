# Flight 11, one mesoscale scene

Starship Flight 11 lifted off from Starbase Pad 1 at 23:23:45 UTC on 13 October 2025. This notebook does not download the bucket. The figure and the inspection table already exist from Milestone 1.

A band-7 pixel at the pad is about 2 km across. Brightness temperature is the Planck temperature of the radiance in that pixel. It is not a flame temperature. The fixed-grid coordinate is where the line of sight meets the ellipsoid. A plume above the surface stays where that line meets the ground. This code does not slide it back onto the pad.

Open the notebook from the repository root. The full walkthrough is `docs/fixed-grid-geometry.md` and `docs/radiance-to-brightness-temperature.md`.


```python
import sys
from pathlib import Path

here = Path.cwd()
root = here if (here / "pyproject.toml").exists() else here.parent
sys.path.insert(0, str(root))

from monitorear.geolocate import LIFTOFF, PAD_LAT, PAD_LON, self_test_geolocation
from monitorear.goes_io import ROOT

print(LIFTOFF.isoformat(), "pad", PAD_LAT, PAD_LON)
self_test_geolocation()

report = ROOT / "outputs" / "m1_inspection.txt"
if report.exists():
    lines = report.read_text().splitlines()
    print("\n".join(lines[:12]))
else:
    print("No inspection report yet. Run scripts/m1_starship_flight11.py")
```

The self-test places the pad on the GRS80 ellipsoid and reads it back. The Flight 11 files carry their own semi-axes; the test only checks the conversion. The report's first frame is the scan that contains the pad before liftoff. The pad pixel on the scan that starts about 10 seconds after liftoff is the 339 K jump, and the next scan peaks near 346 K about 10 km away.
