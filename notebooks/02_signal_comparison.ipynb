# The same statistic at three places

Milestone 2 measured one neighbourhood maximum in three scenes: Pad 1 on the launch evening, Pad 1 on the quiet evening of 25 October 2025, and the strongest CONUS fire pixel at least 100 km from the pad on the scan that overlaps liftoff. It did not choose the detection cut. Milestone 3 later froze that cut at 20 K from these measurements.

The 20 km radius is the score window. The 10–30 km ring is logged beside it and is not the score. This notebook reads the report already written under `outputs/`. It does not list the bucket.


```python
import sys
from pathlib import Path

here = Path.cwd()
root = here if (here / "pyproject.toml").exists() else here.parent
sys.path.insert(0, str(root))

from monitorear.goes_io import ROOT
from monitorear.score import NEIGHBOUR_KM, RING_INNER_KM, RING_OUTER_KM

print(
    f"neighbourhood {NEIGHBOUR_KM:.0f} km, "
    f"ring {RING_INNER_KM:.0f}–{RING_OUTER_KM:.0f} km (logged, not the score)"
)

report = ROOT / "outputs" / "m2_signal.txt"
if report.exists():
    lines = report.read_text().splitlines()
    print("\n".join(lines[:20]))
else:
    print("No signal report yet. Run scripts/m2_signal.py")
```

On the launch evening the 20 km maximum jumped by about 39 K and then about 45 K, then returned near 300 K. On the quiet evening that maximum has a standard deviation of 0.20 K, and the day-to-day background offset is 1.74 K, which is why the score is a jump rather than an absolute brightness temperature. The Georgia fire pixel stayed near 320–343 K across the whole download window. That window is not an ignition time: the first CONUS sample is already hot. The figure is `outputs/m2_signal.png`. The walkthrough is `docs/signal-comparison.md`.
